Bài 1: Yêu cầu. Cước phí Taxi Thông Minh.
Viết một hàm duy nhất smart_taxi_fare(car_type, start_hour, km, wait_minutes) tính và trả về total (tổng tiền phải trả,
kiểu số nguyên, đơn vị VND).
Các quy tắc cần cài đặt:
1. Giá mở cửa (0–0.8 km):
- Xe 4 chỗ: 11,000 VND
- Xe 7 chỗ: 12,000 VND
2. Giá theo km sau 0.8 km:
- Xe 4 chỗ: từ 0.8→20 km ơnh 13,000 VND/km; trên 20 km tính 10,500 VND/km
- Xe 7 chỗ: từ 0.8→20 km ơnh 14,000 VND/km; trên 20 km tính 12,000 VND/km
3. Phí chờ: 5 phút đầu miễn phí; sau đó, mỗi 5 phút (làm tròn lên theo từng block 5 phút) tính 2,000 VND.
4. Phụ thu ban đêm: nếu start_hour thuộc [22,23,0,1,2,3,4,5], cộng +20% trên tiền tạm tính (mở cửa + quãng đường +
chờ).
5. Điều kiện đặc biệt:
- Nếu km > 30 và không phải ban đêm, áp dụng giảm 10% trên tổng tiền.
- Nếu tổng tiền cuối cùng < 50,000 nhưng km > 5, đặt tổng tiền bằng 50,000 VND (mức tối thiểu). 

```python
def smart_taxi_fare(car_type, start_hour, km, wait_minutes):
    total = 0
    a = 0
    if wait_minutes > 5:
        a = ( (wait_minutes // 5) - 1 ) * 2000
    if 22 > start_hour > 5:
        if car_type == 4:
            if 0 < km <= 0.8:
                total == 11000 + a

            if 0.8 < km <= 20:
                total = 11000 + 13000 * (km - 0.8) + a
            if km > 20:
                total = 11000 + 13000 * 19.2 + 10500 * (km-20) + a
            if km > 30:
                total = total * 0.9
            
        if car_type == 7:
            if 0 < km <= 0.8:
                total == 12000 + a

            if 0.8 < km <= 20:
                total = 12000 + 14000 * (km-0.8) + a
            if km > 20:
                total = 12000 + 14000 * 19.2 + 12500 * (km-20) + a
            if km > 30: 
                total = total * 0.9
            
    if total < 50000 and km > 5:
        total == 50000
    if 23>= start_hour >= 22 or 0<= start_hour<=5:
        total = total * 1.2
    return total
car_type, start_hour, km, wait_minutes = map(float, input().split())
print(smart_taxi_fare(car_type, start_hour, km, wait_minutes))





    
    
```

Khuyến khích dùng vòng lặp đơn giản để cộng dồn tổng. 


```python
def cong_don(x: int)-> bool:
    if x < 1:
        return False
    tong = 0
    n = 1
    while tong < x:
        tong += n
        n += 1
    return tong == x
x = int(input())
print(cong_don(x))
```

Bài 25: Yêu cầu: Viết hàm point_in_quadrilateral(quad, p) trả về True nếu điểm p nằm bên trong hoặc trên biên của tứ giác quad,
và False nếu ngược lại.
- Đầu vào:
- quad: danh sách hoặc bộ gồm đúng 4 đỉnh [(x1,y1), (x2,y2), (x3,y3), (x4,y4)] theo thứ tự (tứ giác có thể lồi hoặc lõm,
nhưng không tự cắt).
- p: một điểm (x, y).
- Đầu ra: giá trị boolean.
- Gợi ý: Cài đặt phép kiểm tra điểm trong đa giác tổng quát (ví dụ: ray casting), đồng thời tính cả trường hợp điểm nằm
trên biên là ở trong. 


```python
def point_in_quadrilateral(quad, p):
   
    n = len(quad)
    positive = 0
    negative = 0
    
    for i in range(n):
        
        x1, y1 = quad[i]
        x2, y2 = quad[(i + 1) % n] 
        
        x, y = p
        
        
        d = (x2 - x1) * (y - y1) - (y2 - y1) * (x - x1)
        
        if d > 0:
            positive += 1
        elif d < 0:
            negative += 1
            
    
    return positive == 0 or negative == 0


quad = [(0,0), (4,0), (4,3), (0,3)]
p = (2,1)
print(point_in_quadrilateral(quad, p))
```

Bài 27: Yêu cầu: Viết hàm quadratic_roots_count(a, b, c) trả về số nghiệm thực của phương trình ax^2 + bx + c = 0.
- Đầu vào: các số thực a, b, c.
- Đầu ra: 0, 1, hoặc 2. Nếu phương trình suy biến thành 0 = 0 (tức a == 0, b == 0, c == 0), trả về float('inf').
- Quy tắc:
- Nếu a != 0, dùng biệt thức D = b^2 - 4ac.
- Nếu a == 0 và b != 0, đó là phương trình bậc nhất với đúng một nghiệm thực.
- Nếu a == 0 và b == 0 và c != 0, phương trình vô nghiệm. 


```python
def quadratic_roots_count(a, b, c):
    if a == 0 and b == 0 and c == 0:
        return float('inf')
    elif a == 0 and b == 0 and c != 0:
        return 0
    elif a == 0 and b != 0:
        return 1
    else:
        delta = b**2 - 4*a*c
        if delta > 0:
            return 2
        elif delta == 0:
            return 1
        return 0
a, b, c = map(float, input().split())
print(quadratic_roots_count(a, b, c))
    
```

chấm động, ngược lại trả về 'neither'.
- Đầu vào: chuỗi s.
- Đầu ra: một trong các giá trị {'int', 'float', 'neither'}.
- Quy tắc:
- Cho phép có dấu + hoặc - ở đầu.
- Số nguyên phải chỉ gồm các chữ số và không có dấu chấm thập phân (ví dụ: '100').
- Số thực phải có dấu chấm và có chữ số ở cả hai phía của dấu chấm (ví dụ: '100.0', '-3.14').
- Các chuỗi như '.5', '5.', '1e3' được xem là 'neither' trong bài này.

```python
def detect_number_types(s):
    try:
        int(s)
        return 'int'
    except ValueError:
        try:
            float(s)
            return 'float'
        except ValueError:
            return 'neither'
s = input()
print(detect_number_types(s))
```

Bài 33: Yêu cầu: Viết hàm compute(a, b, op) tính a (+|-||/) b tùy theo op.
- Đầu vào: các số a, b và ký tự op thuộc {'+', '-', '', '/'}.
- Đầu ra: kết quả số học.
- Yêu cầu thêm:
- Nếu op là '/' và b == 0, phát sinh ValueError.
- Nếu op không hợp lệ, phát sinh ValueError. 

```python
def computer(a, b, op):
    if op == "+":
        return a + b
    if op == "-":
        return a - b
    if op == "*":
        return a * b
    if op == "/" and b != 0:
        return a / b
    if op == "/" and b == 0:
        raise ValueError
a, b = map(int, input().split())
op = input()
print(computer(a, b, op))
```

Bài 35: Yêu cầu: Viết hàm divide_with_remainder(dividend, divisor) trả về bộ (q, r) trong đó q là thương nguyên và r là số dư sao
cho dividend = divisorq + r với 0 <= |r| < |divisor| (dùng ngữ nghĩa của Python với // và %).
- Đầu vào: số nguyên dividend, divisor (divisor khác 0).
- Đầu ra: bộ hai số nguyên (q, r).
- Ví dụ: divide_with_remainder(5, 3) -> (1, 2). 


```python
def divide_with_remainder(dividend, divisor):
    q = dividend // divisor
    r = dividend % divisor
    return q, r
dividend, divisor = map(int, input().split())
print(divide_with_remainder(dividend, divisor))
```

Bài 49: Viết chương trình nén chuỗi bằng cách đếm số lượng ký tự liên tiếp giống nhau.
Ví dụ: AAABBB -> 3A3B.
Lưu ý: Chương trình cần phân biệt được chữ hoa, chữ thường, số và khoảng trắng.

```python
def so_ky_tu(n):
    kq = ""
    dem = 1

    for i in range(1, len(n)):
        if n[i] == n[i - 1]:
            dem += 1
        else:
            kq += str(dem) + n[i - 1]
            dem = 1

    kq += str(dem) + n[-1]

    return kq

n = input()
print(so_ky_tu(n))
```

Bài 50: Gọi S(n) là tổng các ước đúng của n (tức là các ước < n).
Một số nguyên dương n được phân loại như sau:
Số hoàn hảo: S(n) = n → in "0"
Số thiếu: S(n) < n → in "-1"
Số dư: S(n) > n → in "1"

Ví dụ
6 → 0
8 → -1
12 → 1

```python
def classify_number(n):
    if n == 1:
        return "-1"  
    
    s = 1 
    i = 2
    while i * i <= n:
        if n % i == 0:
            s += i
            if i * i != n:  
                s += n // i
        i += 1
    
    if s == n:
        return "0"
    elif s < n:
        return "-1"
    else:
        return "1"

n = int(input())
print(classify_number(n))
```
