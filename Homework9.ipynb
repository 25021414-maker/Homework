W9A16:
Cho hai số nguyên dương, n và m (0 < n, m <= 20 ). .Bạn hãy in ra tất cả các chuỗi nhị phân có
n số 0 và m số 1. In theo thứ tự tăng dần trong từ điển 


```python
def generate(n, m, s=""):
    if n == 0 and m == 0:
        print(s)
        return
    
    if n > 0:
        generate(n - 1, m, s + "0")
        
    if m > 0:
        generate(n, m - 1, s + "1")

n, m = map(int, input().split())
generate(n, m)
```

W9A17: Viết hàm tìm dãy con tăng nghiêm ngặt dài nhất trong dãy đầu vào từ bàn phím 


```python
n = list(map(int, input().split()))
max_i = 1
kq = []
for i in range(1, n):
    for j in range(i+1, n):
        if n[j] > n[i]:
            kq.append(j) 
        else:
            len(kq) = 1
if len(kq) > max_i:
    max_i = len(kq)

```

W9A18:
Cho dãy N phần tử số nguyên, hãy đếm số cặp có vị trí khác nhau và có tổng bằng k (số nguyên
dương cho vào từ bàn phím)

```python
n = list(map(int, input().split()))
k = int(input())
dem = 0
kq = []
for i in range(len(n)):
    for j in range(i+1, len(n)):
        if n[i] + n[j] == k:
            dem += 1
            kq.append([i,j])
print(kq, dem)


```

W9A21:
Tìm phần tử lớn thứ k trong dãy n phần tử bằng thuật toán tốt nhất có thể – Tránh dùng sắp xếp
(dùng phân hoạch của quicksort)

```python
n = list(map(int, input().split()))
k = int(input())
for i in range(1, len(n)):
    if n[i] < n[i-1]:
        n[i-1], n[i] = n[i], n[i-1]
print(n[len(n)-k])
```

W9A24:
Nén chuối kí tự bằng cách đếm số lần xuất hiện của các kí tự liên tiếp nhau và ghi lại ngay sau
kí tự đó. VD aaaaaafdhhhhh =>a6f1d1h5 


```python
a = input()
d = {}
for i in a:
    if i in d:
        d[i] += 1
    else:
        d[i] = 1
for keys, values in d.items():
    print(f"{keys}{values}", end="")
```
